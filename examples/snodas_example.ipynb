```python
import snowdat as sn
from datetime import datetime
```

DOWNLOAD BINARY FLAT FILE FROM SNODAS ARCHIVE AND SAVE AS NETCDF

```python
#Define date object using datetime timestamp
date = sn.DatetimeParts(datetime(2025,12,4,0))

#DOWNLOAD AND EXTRACT TAR FILE 
ds = sn.snowdas_dl(date)
```

SAVE RESULT AS COG AND CUSTOM PNG

```python
#define products
products = ['SNOW PRECIP', 'SNOW DEPTH']

bounding_box = {
  "xmin": -80,
  "xmax": -65,
  "ymin": 35,
  "ymax": 48
}


#PROCESS PRODUCTS
for product in products:
    product = product.lower()
    #SAVE AS COG
    sn.snowdas_to_cog(ds, product=product, date=date)
    #SAVE AS PNG
    sn.plot_snowdas(ds, product, date, bounding_box=bounding_box)
```
