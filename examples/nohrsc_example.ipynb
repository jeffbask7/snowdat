```python
import snowdat as sn
from datetime import datetime
```

Download .PNG .TIF or .NC from www.nohrsc.noaa.gov/snowfall_v2/data/

```python
#create a datetime object and pass it to the DatetimeParts class
time_valid = sn.DatetimeParts(datetime(2025,12,5,12))
#set  duration in hours
time_valid.Duration(duration=72)
#OPTIONAL SET EXPORT DIRECTORY BASE PATH
export_dir = '../data'

#FILE TYPE CAN BE '.png', '.tif', '.nc' - DEFAULT IS '.png'
file_path = sn.get_nohrsc(date=time_valid,  file_type='png', export_dir=export_dir)
```

Create Custom Plot from NETCDF File

```python
#OPTIONAL SET BOUNDING AREA TO PLOT
bounding_box = {
  "xmin": -80,
  "xmax": -65,
  "ymin": 35,
  "ymax": 48
}

#PLOT SNOW DATA FILE PATH MUST BE TO .NC FILE
sn.plot_nohrsc_snow(file_path=file_path, bounding_box=bounding_box)
```
